```python
from ultralytics import YOLO

# Load a model
model = YOLO("yolo26n-cls.yaml")  # build a new model from YAML
model = YOLO("yolo26n-cls.pt")  # load a pretrained model (recommended for training)
model = YOLO("yolo26n-cls.yaml").load("yolo26n-cls.pt")  # build from YAML and transfer weights

# Train the model
results = model.train(data="mnist160", epochs=100, imgsz=64)
```

```python
from ultralytics import YOLO

# Load a model
model = YOLO("runs/classify/train/weights/best.pt")  # load a custom model

# Validate the model
metrics = model.val()  # no arguments needed, dataset and settings remembered
print(metrics.top1)  # print results
print(metrics.top5)  # print top5 accuracy
```

```python
model.save("yolo26n-cls-mnist160.pt")  # save the model to a file
```

```python
from ultralytics import YOLO

model = YOLO("runs/classify/train/weights/best.pt")

results = model.predict("bus.jpg")

probs = results[0].probs

if probs is not None:
    top1_id = probs.top1
    top1_conf = float(probs.top1conf)
    top1_name = results[0].names[top1_id]

    print("예측 클래스:", top1_name)
    print("확률:", top1_conf)
else:
    print("분류 결과가 없습니다. 현재 모델이 classify 모델인지 확인하세요.")
```
