```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
```

```python
# CIFAR-10 클래스 이름
CLASS_NAMES = ["airplane", "automobile", "bird", "cat", "deer", "dog", "frog", "horse", "ship", "truck"
]
```

```python
def classify_image(model, image):
    results = model(image)
    probs = results[0].probs
    top1_index = probs.top1
    top1_conf = probs.top1conf.item()
    # predicted_class = CLASS_NAMES[top1_index]
    predicted_class = results[0].names[top1_index]
    print(f"예측 클래스: {predicted_class}")
    print(f"신뢰도: {top1_conf:.4f}")
    
    image = cv2.imread(image)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(6, 6))
    plt.imshow(image)
    plt.title(f"Predicted: {predicted_class} ({top1_conf:.2f})")
    plt.axis('off')
    plt.show()
```

```python
model = YOLO("yolo26n-cls.pt")
image = "airplane.jpg"

# CIFAR-10 학습
# model.train(
#     data="cifar10",
#     epochs=50,
#     imgsz=32,
#     batch=128
# )

classify_image(model, image)
```

```python
def classify_image(model, image):
    results = model(image)
    probs = results[0].probs
    top1_index = probs.top1
    top1_conf = probs.top1conf.item()
    predicted_class = CLASS_NAMES[top1_index]
    print(f"예측 클래스: {predicted_class}")
    print(f"신뢰도: {top1_conf:.4f}")
    
    image = cv2.imread(image)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    plt.figure(figsize=(6, 6))
    plt.imshow(image)
    plt.title(f"Predicted: {predicted_class} ({top1_conf:.2f})")
    plt.axis('off')
    plt.show()
```

```python
model = YOLO("yolo26n-cls.pt")
image = "airplane.jpg"

# CIFAR-10 학습
model.train(
    data="cifar10",
    epochs=50,
    imgsz=32,
    batch=128
)

classify_image(model, image)
```
